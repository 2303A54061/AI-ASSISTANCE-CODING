**SocialMediaTop100.csv**

```python
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

# Number of records
n = 100

# Sample data
names = [f"User_{i}" for i in range(1, n+1)]

platforms = np.random.choice(
    ["Instagram", "Facebook", "Twitter", "TikTok", "LinkedIn"],
    n
)

followers = np.random.randint(100, 1000000, n)
following = np.random.randint(10, 5000, n)
posts = np.random.randint(1, 2000, n)
engagement_rate = np.round(np.random.uniform(0.1, 15.0, n), 2)

# Create DataFrame
social_df = pd.DataFrame({
    "PROFILE": names,
    "Platform": platforms,
    "Followers": followers,
    "Following": following,
    "Posts": posts,
    "Engagement_Rate": engagement_rate
})

# Save as CSV
social_df.to_csv("SocialMediaTop100.csv", index=False)

print("CSV file created successfully!")

# Show sample
social_df.head()
```

**weather_forecast_data.csv**

```python
import pandas as pd
import numpy as np

# Reproducibility
np.random.seed(42)

# Number of records
n = 100

# Sample cities (you can change or extend)
cities = np.random.choice(
    ["Hyderabad", "Chennai", "Bangalore", "Delhi", "Mumbai"],
    n
)

# Generate synthetic weather data
temperature = np.round(np.random.uniform(15, 45, n), 1)   # °C
humidity = np.random.randint(20, 100, n)                  # %
wind_speed = np.round(np.random.uniform(0, 25, n), 1)     # km/h
pressure = np.random.randint(980, 1050, n)                # hPa

conditions = np.random.choice(
    ["Sunny", "Cloudy", "Rainy", "Stormy", "Foggy"],
    n
)

# Create DataFrame
weather_df = pd.DataFrame({
    "City": cities,
    "Temperature_C": temperature,
    "Humidity_%": humidity,
    "Wind_Speed_kmph": wind_speed,
    "Pressure_hPa": pressure,
    "Condition": conditions
})

# Save CSV
weather_df.to_csv("weather_forecast_data.csv", index=False)

print("weather_forecast_data.csv created successfully!")

# Show sample
weather_df.head()
```

```python
# Task 1
import statistics

def analyze_numbers(numbers):
    """
    Calculates and prints the mean, minimum, and maximum of a list of numbers.

    Args:
        numbers (list): A list of numerical values.
    """
    if not numbers:
        print("The list is empty. Cannot calculate statistics.")
        return

    mean_value = statistics.mean(numbers)
    min_value = min(numbers)
    max_value = max(numbers)

    print(f"List of numbers: {numbers}")
    print(f"Mean: {mean_value}")
    print(f"Minimum: {min_value}")
    print(f"Maximum: {max_value}")

# Run the function on the specified list
my_list = [12, 7, 9, 21, 5]
analyze_numbers(my_list)
```

```python
import pandas as pd
import numpy as np

# Load dataset
data = pd.read_csv('weather_forecast_data.csv')

# 1. Create Date column (since dataset doesn't have it)
data['Date'] = pd.date_range(start="2025-01-01", periods=len(data))

# 2. Convert to datetime
data['Date'] = pd.to_datetime(data['Date'])

# 3. Handle missing values (safe check)
for col in ['Temperature_C', 'Humidity_%']:
    if col in data.columns:
        data[col] = data[col].fillna(data[col].median())

# 4. Season feature engineering
def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
    elif month in [6, 7, 8]:
        return 'Summer'
    else:
        return 'Autumn'

data['Season'] = data['Date'].dt.month.apply(get_season)

# 5. Week number feature
data['Week_Number'] = data['Date'].dt.isocalendar().week

# 6. Normalize temperature (instead of rainfall)
if 'Temperature_C' in data.columns:
    data['Temp_Normalized'] = (
        (data['Temperature_C'] - data['Temperature_C'].min()) /
        (data['Temperature_C'].max() - data['Temperature_C'].min())
    )

# Show result
data.head()
```

```python
import pandas as pd
from sklearn.preprocessing import MultiLabelBinarizer, MinMaxScaler

# Load the dataset
movies = pd.read_csv('IMDB-Movie-Data.csv')

# Remove duplicate movie entries based on Title and Year
movies = movies.drop_duplicates(subset=['Title', 'Year'])

# Handle missing values in 'Rating' and 'Genre'
if 'Rating' in movies.columns:
    movies['Rating'].fillna(movies['Rating'].median(), inplace=True)
if 'Genre' in movies.columns:
    movies['Genre'].fillna('Unknown', inplace=True)

# Encode 'Genre' as multi-hot (one-hot for each genre)
if 'Genre' in movies.columns:
    mlb = MultiLabelBinarizer()
    genre_lists = movies['Genre'].str.split(',')
    genre_encoded = mlb.fit_transform(genre_lists)
    genre_df = pd.DataFrame(genre_encoded, columns=mlb.classes_)
    movies = pd.concat([movies, genre_df], axis=1)

# Encode 'Language' if present
if 'Language' in movies.columns:
    movies['Language'] = movies['Language'].astype(str)
    movies['Language_encoded'] = movies['Language'].astype('category').cat.codes

# Scale 'Rating' between 0 and 1
if 'Rating' in movies.columns:
    scaler = MinMaxScaler()
    movies['Rating_scaled'] = scaler.fit_transform(movies[['Rating']])

# Save the cleaned and processed dataframe to a new CSV
movies.to_csv('IMDB-Movie-Data-processed.csv', index=False)

# Show the first few rows
print(movies.head())


# Explanation in simple words:
# 1. We load the movie dataset from a CSV file.
# 2. We remove duplicate movie entries based on Title and Year.
# 3. We handle missing values in 'Rating' and 'Genre'.
# 4. We encode 'Genre' as multi-hot (one-hot for each genre).
# 5. We encode 'Language' as numeric codes.
# 6. We scale 'Rating' to be between 0 and 1 using MinMaxScaler.
```
